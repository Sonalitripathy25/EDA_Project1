## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# 1. Import necessary libraries
import pandas as pd
import numpy as np

# 2. Load the dataset
df = pd.read_csv(file_path)  # Replace with your file path

# 3. Convert 'Date' column to datetime format
df['Date'] = pd.to_datetime(df['Date'], format='%b-%y')

# 4. Sort the dataset by date (important for time series)
df = df.sort_values(by='Date').reset_index(drop=True)

# 5. Remove duplicate rows
df = df.drop_duplicates()

# 6. Handle missing values
df = df.dropna(how='all')                  # Drop rows with all values missing
# df.fillna(method='ffill', inplace=True)    # Forward-fill remaining missing values # There are no missing values as shown in the previous steps.

# 7. Create time-based features
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Quarter'] = df['Date'].dt.quarter

# 8. Create derived features
df['Price_Change'] = df['Close'] - df['Open']
df['Price_Range'] = df['High'] - df['Low']
# Volatility could be calculated as (High - Low) / Close or Open depending on preference and
# domain knowledge. Let's use Close for now.
df['Volatility'] = (df['High'] - df['Low']) / df['Close']


# 9. Convert price columns to numeric (in case of any errors)
price_columns = ['Open', 'High', 'Low', 'Close']
df[price_columns] = df[price_columns].apply(pd.to_numeric, errors='coerce')

# 10. Final check
print("Dataset is analysis-ready. Summary:")
print(df.info())
print(df.head())
````

There is no significant difference in the mean closing prices of Yes Bank before and after 2018.
⇒ μ₁ = μ₂

Alternate Hypothesis (H₁):
There is a significant difference in the mean closing prices of Yes Bank before and after 2018.
⇒ μ₁ ≠ μ₂

Where:

μ₁ = mean closing price before 2018

μ₂ = mean closing price after 2018

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
# test Hypothesis 1 and obtain the p‑value (Welch’s two‑sample t‑test: mean Close before 2018 vs. from 2018 onward). It prints sample sizes, means, t‑statistic, p‑value, effect size, and a 95% CI for the mean difference.

# Hypothesis 1:
# H0: mean Close (pre-2018) = mean Close (2018 and after)
# H1: mean Close (pre-2018) ≠ mean Close (2018 and after)

import pandas as pd
import numpy as np
from scipy import stats

# Ensure proper dtypes and ordering
# Assuming 'df' is already loaded from the previous steps, ensure 'Date' is datetime
df['Date'] = pd.to_datetime(df['Date'], format='%b-%y') # Convert 'Date' column to datetime with specified format
df = df.sort_values('Date')

# Split groups
pre = df.loc[df['Date'] < pd.Timestamp('2018-01-01'), 'Close'].astype(float).dropna()
post = df.loc[df['Date'] >= pd.Timestamp('2018-01-01'), 'Close'].astype(float).dropna()

print(f"Pre-2018: n={pre.size}, mean={pre.mean():.3f}, std={pre.std(ddof=1):.3f}")
print(f"2018+   : n={post.size}, mean={post.mean():.3f}, std={post.std(ddof=1):.3f}")

# Check if both groups have data before performing the test
if pre.size > 1 and post.size > 1:
    # Welch's t-test (does not assume equal variances)
    t_stat, p_val = stats.ttest_ind(pre, post, equal_var=False, nan_policy='omit')
    print(f"\nWelch t-test: t={t_stat:.3f}, p-value={p_val:.6f}")

    # Effect size (Cohen's d using pooled SD with unbiased estimator g reported)
    n1, n2 = pre.size, post.size
    s1, s2 = pre.std(ddof=1), post.std(ddof=1)
    pooled_sd = np.sqrt(((n1-1)*s1**2 + (n2-1)*s2**2) / (n1+n2-2)) if (n1+n2-2) > 0 else np.nan
    cohens_d = (pre.mean() - post.mean()) / pooled_sd if pooled_sd > 0 else np.nan
    # Hedges' g correction
    J = 1 - (3 / (4*(n1+n2) - 9)) if (n1+n2) > 2 else 1.0
    hedges_g = J * cohens_d
    print(f"Effect size: Cohen's d={cohens_d:.3f}, Hedges' g={hedges_g:.3f}")

    # 95% CI for mean difference using Welch-Satterthwaite df
    mean_diff = pre.mean() - post.mean()
    se = np.sqrt((s1**2 / n1) + (s2**2 / n2)) if n1 > 0 and n2 > 0 else np.nan
    # Welch-Satterthwaite degrees of freedom
    df_welch = (s1**2/n1 + s2**2/n2)**2 / ((s1**2/n1)**2/(n1-1) + (s2**2/n2)**2/(n2-1)) if n1 > 1 and n2 > 1 else np.nan
    t_crit = stats.t.ppf(0.975, df_welch) if df_welch > 0 else np.nan
    ci_low, ci_high = mean_diff - t_crit*se, mean_diff + t_crit*se if not np.isnan(t_crit) and not np.isnan(se) else (np.nan, np.nan)
    print(f"95% CI for (pre - post) mean difference: [{ci_low:.3f}, {ci_high:.3f}]")

    # Simple decision at alpha = 0.05
    alpha = 0.05
    if p_val < alpha:
        print("\nDecision: Reject H0 at α=0.05. The mean monthly closing price differs before vs. after 2018.")
    else:
        print("\nDecision: Fail to reject H0 at α=0.05. No significant difference detected.")
else:
    print("\nNot enough data in one or both groups to perform the t-test.")
````

````python
# Perform Statistical Test to obtain P-Value
import pandas as pd
from scipy.stats import ttest_ind
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv(file_path) # Use the file_path variable

# Convert 'Date' column to datetime if not already
df['Date'] = pd.to_datetime(df['Date'], format='%b-%y') # Specify the format

# Extract year from date
df['Year'] = df['Date'].dt.year

# Calculate Volatility (High - Low)
df['Volatility'] = df['High'] - df['Low']

# Filter data before and after 2018 based on Volatility
before_2018 = df[df['Year'] < 2018]['Volatility'].dropna()
after_2018 = df[df['Year'] >= 2018]['Volatility'].dropna()

# Perform independent t-test (two-tailed initially, but we can adjust based on hypothesis)
# For the hypothesis H1: μafter_2018 > μbefore_2018, we should use a one-tailed test.
# The ttest_ind function in scipy performs a two-tailed test by default.
# To get the p-value for a one-tailed test (greater than), we can divide the two-tailed p-value by 2
# if the t-statistic is in the expected direction (positive in this case, as we expect after_2018 mean to be greater).
t_stat, p_val_two_tailed = ttest_ind(before_2018, after_2018, equal_var=False)

# Check the direction of the difference and calculate the one-tailed p-value
if t_stat > 0: # If mean of before_2018 is greater than after_2018, opposite of hypothesis
    p_val_one_tailed = 1 - p_val_two_tailed / 2
else: # If mean of after_2018 is greater than before_2018, in the direction of hypothesis
    p_val_one_tailed = p_val_two_tailed / 2


# Print test result
print("T-Statistic:", t_stat)
print("Two-tailed P-Value:", p_val_two_tailed)
print("One-tailed P-Value (after_2018 > before_2018):", p_val_one_tailed)


# Interpret the result for the one-tailed hypothesis
alpha = 0.05
if p_val_one_tailed < alpha:
    print("\nResult: Reject the null hypothesis. The average volatility is significantly higher after 2018.")
else:
    print("\nResult: Fail to reject the null hypothesis. There is no significant evidence that the average volatility is higher after 2018.")
````

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
# 1. Setup and Required Imports
import pandas as pd
import numpy as np
from scipy.stats import zscore
import matplotlib.pyplot as plt
import seaborn as sns

# Load your dataset
df = pd.read_csv(file_path)
````

````python
#  2. Detecting Outliers Using IQR Method
def detect_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return df[(df[column] < lower_bound) | (df[column] > upper_bound)]

# Example: Detect outliers in 'Close Price'
outliers_close = detect_outliers_iqr(df, 'Close')
print(f"Outliers in Close Price: {len(outliers_close)}")
````

````python
# 3. Treating Outliers: Capping Method
def cap_outliers(df, column):
    lower = df[column].quantile(0.01)
    upper = df[column].quantile(0.99)
    df[column] = np.where(df[column] < lower, lower, df[column])
    df[column] = np.where(df[column] > upper, upper, df[column])
    return df

# Apply capping to selected numeric columns
cols_to_cap = ['Open', 'Close', 'High', 'Low']
for col in cols_to_cap:
    df = cap_outliers(df, col)
````

````python
#  4. Z-Score Based Filtering (Alternative Method)
# Apply Z-score only to normally distributed columns
numeric_cols = ['Open', 'Close', 'High', 'Low']
z_scores = np.abs(zscore(df[numeric_cols]))
df_z_filtered = df[(z_scores < 3).all(axis=1)]  # Keep rows with all Z < 3
print(f"Rows after Z-Score filtering: {df_z_filtered.shape[0]}")
````

````python
# ML Model - 3 Implementation

# Fit the Algorithm

# Predict on the model

# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import GradientBoostingRegressor # Using Gradient Boosting Regressor as ML Model 3
from sklearn.metrics import mean_squared_error, r2_score # Using regression metrics
import matplotlib.pyplot as plt
import seaborn as sns

# Assuming X_train, X_test, y_train, and y_test are already defined from the data splitting step

# Step 1: Define the Gradient Boosting Regressor model
gbr = GradientBoostingRegressor(random_state=42)

# Step 2: Define the hyperparameters grid for Gradient Boosting
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [3, 4, 5],
    'learning_rate': [0.01, 0.1, 0.2],
    'min_samples_split': [2, 5, 10]
}

# Step 3: Use GridSearchCV for hyperparameter tuning
grid_search = GridSearchCV(estimator=gbr,
                           param_grid=param_grid,
                           cv=5, # Using 5-fold cross-validation
                           n_jobs=-1, # Use all available cores
                           verbose=1, # Print progress
                           scoring='neg_mean_squared_error') # Scoring for regression

# Step 4: Fit the model on the training data
grid_search.fit(X_train, y_train)

# Step 5: Predict using the best estimator found by GridSearchCV
best_gbr = grid_search.best_estimator_
y_pred = best_gbr.predict(X_test)

# Step 6: Evaluate the model using regression metrics
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("Best Hyperparameters:", grid_search.best_params_)
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R-squared Score (R2): {r2:.4f}")

# Optional: Plotting Actual vs Predicted values
# plt.figure(figsize=(10, 6))
# plt.scatter(y_test, y_pred, alpha=0.5)
# plt.xlabel("Actual Close Price")
# plt.ylabel("Predicted Close Price")
# plt.title("Actual vs Predicted Close Price (Gradient Boosting Regressor)")
# plt.show()
````

1. Accuracy
Definition:
The ratio of correctly predicted instances to the total instances.

Why It Matters:

A general indicator of model performance.

Useful when the dataset is balanced.

Helps measure how often the model is right overall.

✅ Business Impact:

High accuracy means fewer incorrect decisions in stock movement prediction, improving strategic decision-making.

2. Precision
Definition:
The ratio of correctly predicted positive observations to the total predicted positives.
Formula: TP / (TP + FP)

Why It Matters:

Crucial when the cost of a false positive is high.

Example: Predicting a price rise when it will actually drop can cause financial loss.

✅ Business Impact:

Higher precision reduces risky investments based on wrong predictions.

Builds investor confidence in algorithmic trading.

3. Recall (Sensitivity)
Definition:
The ratio of correctly predicted positives to all actual positives.
Formula: TP / (TP + FN)

Why It Matters:

Important when missing a positive case is costly.

Example: Failing to identify a profitable opportunity.

✅ Business Impact:

Higher recall means fewer missed profit opportunities.

Helps portfolio managers to maximize gains from timely trades.

4. F1 Score
Definition:
The harmonic mean of precision and recall.

Why It Matters:

Useful when you want a balance between precision and recall.

More reliable than accuracy for imbalanced datasets.

✅ Business Impact:

Ensures consistency and reliability of predictions.

Reduces the chance of overfitting to either high precision or recall, aligning with real-world financial risk.
