## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Import Required Libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

#  Step 1: Drop Irrelevant Columns
# df.drop(columns=['ID', 'Customer_ID', 'Name', 'SSN'], inplace=True) # These columns were already dropped

#  Step 2: Handle Missing Values

# Fill numeric columns with median
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill categorical columns with mode
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

#  Step 3: Convert 'Credit_History_Age' to numeric (total months)
# The Credit_History_Age is already in numeric format


df['Credit_History_Age'].fillna(df['Credit_History_Age'].median(), inplace=True)


#  Step 4: Encode Categorical Variables
le = LabelEncoder()
# Select only object type columns for encoding, excluding those already encoded
cat_cols_to_encode = df.select_dtypes(include='object').columns
for col in cat_cols_to_encode:
    df[col] = le.fit_transform(df[col])


#  Step 5: Final Check
print(" Final Dataset Shape:", df.shape)
print(" Remaining Missing Values:", df.isnull().sum().sum())
display(df.head())
````

````python
# Chart - 1 visualization code
# Chart 1 – Credit Score Distribution
import seaborn as sns
import matplotlib.pyplot as plt

# Set the figure size
plt.figure(figsize=(8, 6))

# Define the mapping from numerical to string labels
credit_score_mapping = {0: 'Good', 1: 'Poor', 2: 'Standard'}

# Plot the count of each Credit Score category
# Use the numerical values for plotting and the mapping for labels
ax = sns.countplot(data=df, x='Credit_Score', palette='Set2', hue='Credit_Score', legend=False)

# Set the x-axis ticks and labels
# Get the current tick locations
tick_locs = ax.get_xticks()
# Map the numerical tick locations to the string labels
tick_labels = [credit_score_mapping[int(loc)] for loc in tick_locs]

ax.set_xticks(tick_locs)
ax.set_xticklabels(tick_labels)

# Add titles and labels
plt.title('Credit Score Distribution')
plt.xlabel('Credit Score Category')
plt.ylabel('Number of Customers')

# Add percentage labels
total = len(df)
for p in ax.patches:
    percentage = '{:.1f}%'.format(100 * p.get_height() / total)
    x = p.get_x() + p.get_width() / 2 - 0.05
    y = p.get_height()
    ax.annotate(percentage, (x, y), ha='center', va='bottom')


# Show the plot
plt.show()
````
